```python
import os
import tiktoken
from datasets import load_dataset
```

```python
def count_py_files(directory):
    count = 0
    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.endswith('.py'):
                count += 1
    return count

def count_py_files_and_tokens(directory):
    py_file_count = 0
    total_tokens = 0
    encoding = tiktoken.get_encoding("cl100k_base")  # 適用於 GPT-4 / GPT-3.5

    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.endswith('.py'):
                py_file_count += 1
                file_path = os.path.join(root, file)
                try:
                    with open(file_path, 'r', encoding='utf-8') as f:
                        content = f.read()
                    token_count = len(encoding.encode(content))
                    total_tokens += token_count
                    # print(f"[{file_path}] → {token_count} tokens")
                except Exception as e:
                    print(f"[{file_path}] → 無法讀取 ({e})")

    return py_file_count, total_tokens
```

```python
dataset = load_dataset("princeton-nlp/SWE-bench_Lite", split="test")
```

```python
total_tokens = 0

for task in dataset:
    instance_id = task['instance_id']
    path = f'/home/tweichuan/project/codebases/{instance_id}'
    if os.path.isdir(path):
        py_count, tokens = count_py_files_and_tokens(path)
        print(f"Total tokens in '{instance_id}': {tokens}")
        total_tokens += tokens
    else:
        print("Invalid directory path.")
```
